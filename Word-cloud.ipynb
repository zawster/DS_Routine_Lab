```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud
import numpy as np
```

```python
# Define the words and frequencies
word_frequencies = {
    "AI Engineering": 15, "PyTorch": 14, "Keras": 10, "TensorFlow": 15, "Pandas": 10,
    "LangChain": 12, "A/B Testing": 5, "LoRA": 4, "MLOps": 11, "FastAPI": 9,
    "Huggingface": 11, "Docker": 11, "Kubernetes": 5, "AWS": 8, "Azure": 8, "GCP": 6, "Python": 15
}

wordcloud = WordCloud(
    width=1584,
    height=396,
    background_color="navy",  # Set the background to navy blue
    colormap="cool",  # Color scheme for words
    max_words=100,
    contour_color='white',
    contour_width=2,
).generate_from_frequencies(word_frequencies)

plt.figure(figsize=(15.84, 3.96))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")  # Hide axes
plt.show()
```
